```python
#COMPARISON PLOT
import plotly.graph_objects as go
import matplotlib.pyplot as plt


categories = ['ACC', 'PRE', 'SEN', 'SPE', 'MCC']

fig = go.Figure()

fig.add_trace(go.Scatterpolar(
    r=[81.48, 82.41, 82.61, 80.59, 63.05],
    theta=categories,
    fill='toself',
    name='ACP-DL'
))
fig.add_trace(go.Scatterpolar(
    r=[0.80, 79, 83, 80, 62.30],
    theta=categories,
    fill='toself',
    name='AntiCP2.0'
))
fig.add_trace(go.Scatterpolar(
    r=[84, 83.14, 83.12, 84.32, 66],
    theta=categories,
    fill='toself',
    name='DeepACP'
))
fig.add_trace(go.Scatterpolar(
    r=[82.03, 84.14, 86.98, 83.26, 64.71],
    theta=categories,
    fill='toself',
    name='ACP-DA'
))
fig.add_trace(go.Scatterpolar(
    r=[86.48, 88.63, 84.32, 88.73, 73.19],
    theta=categories,
    fill='toself',
    name='ACP-ADA'
))
fig.add_trace(go.Scatterpolar(
    r=[95.10, 96.20, 95.33, 94.57, 88.70],
    theta=categories,
    fill='toself',
    name='GAN-ML'
))

fig.update_layout(
    polar=dict(
        radialaxis=dict(
            visible=True,
            range=[0, 100]
        )),
    showlegend=True,
    width=600,  # Set the width of the figure
    height=400  # Set the height of the figure
)

fig.show()
plt.savefig("comparison.png")
```
