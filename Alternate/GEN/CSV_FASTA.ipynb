```python
import csv

def fasta_to_csv(fasta_file, csv_file):
    # Open FASTA file for reading
    with open(fasta_file, 'r') as fasta:
        sequences = []
        sequence_id = None
        sequence = ''

        # Read through each line in the file
        for line in fasta:
            line = line.strip()
            if line.startswith('>'):  # New sequence identifier
                # Store previous sequence if exists
                if sequence_id and sequence:
                    sequences.append((sequence_id, sequence))
                    sequence = ''

                # Get new sequence identifier
                sequence_id = line[1:]  # Remove '>' character
            else:  # Sequence line
                sequence += line

        # Append the last sequence
        if sequence_id and sequence:
            sequences.append((sequence_id, sequence))

    # Write to CSV file
    with open(csv_file, 'w', newline='') as csv_file:
        writer = csv.writer(csv_file)
        # Write header
        writer.writerow(['Sequence_ID', 'Sequence'])
        # Write sequences
        writer.writerows(sequences)

    print(f'Conversion completed. {len(sequences)} sequences written to {csv_file}')

# Example usage:
fasta_file = 'gen_acp.txt'  # Replace with your FASTA file path
csv_file = 'gen.csv'  # Replace with desired CSV output file path
fasta_to_csv(fasta_file, csv_file)
```
