```python
def read_fasta_file(filename):
    sequences = {}
    current_id = None
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            if line.startswith('>'):
                current_id = line[1:]
                sequences[current_id] = ''
            else:
                sequences[current_id] += line
    return sequences

def pad_sequences(sequences):
    max_length = max(len(seq) for seq in sequences.values())
    padded_sequences = {}
    for seq_id, seq in sequences.items():
        padded_seq = seq + '-' * (max_length - len(seq))
        padded_sequences[seq_id] = padded_seq
    return padded_sequences

def write_fasta_file(filename, sequences):
    with open(filename, 'w') as file:
        for seq_id, seq in sequences.items():
            file.write(f'>{seq_id}\n{seq}\n')

# Example usage:
fasta_file = 'non_acp_neg_train.fasta'
sequences = read_fasta_file(fasta_file)
padded_sequences = pad_sequences(sequences)
write_fasta_file('neg_non-acp_pad.fasta', padded_sequences)
print('New FASTA file with padded sequences created.')
```
