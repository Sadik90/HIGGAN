```python
from Bio import SeqIO

def filter_sequences(input_file, output_file):
    with open(input_file, "r") as input_handle, open(output_file, "w") as output_handle:
        # Iterate over each sequence in the input FASTA file
        for record in SeqIO.parse(input_handle, "fasta"):
            # Check the length of the sequence
            if len(record.seq) < 50:
                # Write sequences with length less than 50 to the output FASTA file
                SeqIO.write(record, output_handle, "fasta")

# Input and output file paths
input_file = "AMP_database.fasta"
output_file = "outputAMP_50.fasta"

# Call the function to filter sequences
filter_sequences(input_file, output_file)

print("Sequences with length less than 50 have been filtered and saved to output.fasta")
```
