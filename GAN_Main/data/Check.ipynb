```python

import pandas as pd
from matplotlib_venn import venn2
import matplotlib.pyplot as plt

# Load train.csv and test.csv
train_df = pd.read_csv('train/Seq_X_train_all.csv')
test_df = pd.read_csv('test/Seq_X_test_all.csv')

# Get the sequences from both DataFrames
train_sequences = set(train_df['Sequence'])
test_sequences = set(test_df['Sequence'])

# Find common sequences
common_sequences = train_sequences.intersection(test_sequences)

# Get the number of common sequences
num_common_sequences = len(common_sequences)

# Print the number of common sequences
print(f'Number of common sequences: {num_common_sequences}')

# Print common sequences
print('Common Sequences:')
for seq in common_sequences:
    print(seq)

# Create a Venn diagram
plt.figure(figsize=(8, 6))
venn2([train_sequences, test_sequences], ('Train Sequences', 'Test Sequences'))
plt.title(f'Common Sequences: {num_common_sequences}')
plt.show()
```

```python
import pandas as pd
from matplotlib_venn import venn2
import matplotlib.pyplot as plt

# Load train.csv and test.csv
train_df = pd.read_csv('train/Seq_X_train_all.csv')
test_df = pd.read_csv('test/Seq_X_test_all.csv')

# Get the sequences from both DataFrames
train_sequences = train_df['Sequence']
test_sequences = test_df['Sequence']

# Find duplicate sequences in both train and test datasets
num_duplicates_train = train_sequences.duplicated().sum()
num_duplicates_test = test_sequences.duplicated().sum()

# Print the number of duplicates in each dataset
print(f'Number of duplicate sequences in train dataset: {num_duplicates_train}')
print(f'Number of duplicate sequences in test dataset: {num_duplicates_test}')

# Create a Venn diagram
plt.figure(figsize=(8, 6))
venn2([set(train_sequences), set(test_sequences)], ('Train Sequences', 'Test Sequences'))
plt.title(f'Common Sequences: {num_common_sequences}')
plt.show()
```

```python
import pandas as pd

# Load train.csv and test.csv
train_df = pd.read_csv('train/Seq_X_train_all.csv')
test_df = pd.read_csv('test/Seq_X_test_all.csv')

# Remove duplicate sequences
train_df = train_df.drop_duplicates(subset=['Sequence'])
test_df = test_df.drop_duplicates(subset=['Sequence'])

# Save the modified DataFrames back to CSV files
train_df.to_csv('train/Seq_X_train_all.csv', index=False)
test_df.to_csv('test/Seq_X_test_all.csv', index=False)
```

```python
import pandas as pd

# Load train.csv and test.csv
train_df = pd.read_csv('train/Seq_X_train_all.csv')
test_df = pd.read_csv('test/Seq_X_test_all.csv')

# Get the sequences from train and test DataFrames
train_sequences = set(train_df['Sequence'])
test_sequences = set(test_df['Sequence'])

# Find sequences that are both in train and test sets
common_sequences = train_sequences.intersection(test_sequences)

# Remove common sequences from the test set
test_df = test_df[~test_df['Sequence'].isin(common_sequences)]

# Save the modified DataFrames back to CSV files
train_df.to_csv('train/Seq_X_train_all.csv', index=False)
test_df.to_csv('test/Seq_X_test_all.csv', index=False)
```
